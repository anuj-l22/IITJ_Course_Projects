# Q1

```python
import numpy as np
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score

y1 = np.load("labels_main.npy")   # from original main.py (PyG)
y2 = np.load("labels_main2.npy")  # from original main2.py (raw)

if len(y1) != len(y2):
    n = min(len(y1), len(y2))
    print(f"[warn] length mismatch {len(y1)} vs {len(y2)}; truncating to {n}")
    y1, y2 = y1[:n], y2[:n]

ari = adjusted_rand_score(y1, y2)
nmi = normalized_mutual_info_score(y1, y2)

print("=== Q1: PyG vs Raw (original implementations) ===")
print("Adjusted Rand Index (ARI):", ari)
print("Normalized Mutual Information (NMI):", nmi)
print("#unique groups (PyG, Raw):", len(np.unique(y1)), len(np.unique(y2)))
print("Identical partitions? ->", "YES" if (ari == 1.0 and nmi == 1.0) else "NO")
```
